# Customer Data Analysis using Python

This project focuses on cleaning, analyzing, and visualizing customer/sales data to extract meaningful business insights.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
```

## Load Dataset
The dataset is loaded and inspected to understand its structure and basic properties.


```python
from google.colab import files
uploaded = files.upload()
```

```python
df = pd.read_csv(list(uploaded.keys())[0], encoding='latin1')
df.head()
```

## Dataset Overview
We examine the dataset dimensions, data types, summary statistics, and missing values.


```python
df.info()
df.describe()
df.isnull().sum()
```

## Data Cleaning
Handling missing values and removing duplicate records to improve data quality.
Since there is no missing values we no need to perform it.

```python
df = df.drop_duplicates()
df = df.dropna()
```

## Exploratory Data Analysis (EDA)


```python
plt.figure(figsize=(6,4))
sns.histplot(df.select_dtypes(include=np.number).iloc[:,0], kde=True)
plt.title("Distribution of a Numerical Feature")
plt.show()
```

## Insights
- Cleaned data shows improved consistency
- Visualizations help understand distribution patterns
- Useful trends identified for business decision-making


## Conclusion
This project demonstrates the importance of data cleaning and exploratory data analysis in extracting meaningful insights from raw data.
